## 宋史

```python
import os
from bs4 import BeautifulSoup

# Define the base directory
base_directory = '/content/drive/MyDrive/NTU/研究/LLM Hackathon/宋史_列傳/'

# Check if Google Drive is mounted
if not os.path.exists('/content/drive'):
    print("Error: Google Drive not mounted. Please mount your Google Drive first.")
elif not os.path.exists(base_directory):
    print(f"Error: Base directory '{base_directory}' not found. Please ensure the path is correct.")
else:
    print(f"Processing files in subdirectories under: {base_directory}")

    # Iterate through volumes from 242 to 496
    for volume_num in range(242, 497):
        volume_dir_name = f'宋史_卷{volume_num}'
        current_directory_path = os.path.join(base_directory, volume_dir_name)

        if not os.path.exists(current_directory_path):
            print(f"Warning: Directory '{current_directory_path}' not found. Skipping.")
            continue

        print(f"\nProcessing volume directory: {current_directory_path}")
        # Iterate over all files in the current volume directory
        for filename in os.listdir(current_directory_path):
            if filename.endswith('.html') or filename.endswith('.htm'):
                html_filepath = os.path.join(current_directory_path, filename)
                txt_filename = os.path.splitext(filename)[0] + '.txt'
                txt_filepath = os.path.join(current_directory_path, txt_filename)

                try:
                    with open(html_filepath, 'r', encoding='utf-8') as f:
                        html_content = f.read()

                    # Parse HTML and extract text
                    soup = BeautifulSoup(html_content, 'html.parser')
                    # Get text, using a space as separator and stripping extra whitespace
                    text_content = soup.get_text(separator=' ', strip=True)

                    # Save the text content to a new .txt file
                    with open(txt_filepath, 'w', encoding='utf-8') as f:
                        f.write(text_content)
                    print(f"Converted '{filename}' to '{txt_filename}' in '{volume_dir_name}'")
                except Exception as e:
                    print(f"Error processing '{filename}' in '{volume_dir_name}': {e}")
    print("\nConversion of all specified volumes complete.")
```

## 明史

```python
# ============================================================
# 明史 卷113～卷332
# HTML → TXT
# TXT 儲存於各自原本的資料夾
# ============================================================

from google.colab import drive
from pathlib import Path
from bs4 import BeautifulSoup

# ============================================================
# 1. 掛載 Google Drive
# ============================================================

drive.mount('/content/drive')

# ============================================================
# 2. 基本路徑
# ============================================================

base_path = Path(
    "/content/drive/MyDrive/NTU/研究/LLM Hackathon/明史_列傳"
)

START_VOLUME = 113
END_VOLUME = 332

# ============================================================
# 3. 統計
# ============================================================

total_folders = 0
missing_folders = 0
total_html = 0
success_count = 0
error_count = 0

# ============================================================
# 4. 逐一處理「明史_卷113」～「明史_卷332」
# ============================================================

for volume in range(START_VOLUME, END_VOLUME + 1):

    # --------------------------------------------------------
    # 建立資料夾路徑
    # --------------------------------------------------------

    folder = base_path / f"明史_卷{volume}"

    # --------------------------------------------------------
    # 檢查資料夾是否存在
    # --------------------------------------------------------

    if not folder.exists():
        print(f"⚠ 找不到資料夾：{folder}")
        missing_folders += 1
        continue

    total_folders += 1

    print("\n" + "=" * 70)
    print(f"處理：明史_卷{volume}")
    print("=" * 70)

    # --------------------------------------------------------
    # 找出該資料夾內的 HTML
    # --------------------------------------------------------

    html_files = list(folder.glob("*.html")) + list(folder.glob("*.htm"))

    print(f"找到 {len(html_files)} 個 HTML 檔案")

    total_html += len(html_files)

    # --------------------------------------------------------
    # 逐一轉換
    # --------------------------------------------------------

    for html_file in sorted(html_files):

        try:

            # =================================================
            # 讀取 HTML
            # =================================================

            with open(
                html_file,
                "r",
                encoding="utf-8",
                errors="replace"
            ) as f:
                html_content = f.read()

            # =================================================
            # BeautifulSoup 解析 HTML
            # =================================================

            soup = BeautifulSoup(
                html_content,
                "html.parser"
            )

            # =================================================
            # 移除 script / style / noscript
            # =================================================

            for tag in soup(["script", "style", "noscript"]):
                tag.decompose()

            # =================================================
            # 取得純文字
            # =================================================

            text = soup.get_text(separator="\n")

            # =================================================
            # 清理文字
            # =================================================

            text = text.replace("\r\n", "\n")
            text = text.replace("\r", "\n")

            # 每行去除前後空白
            lines = [
                line.strip()
                for line in text.split("\n")
            ]

            # -------------------------------------------------
            # 移除連續空白行
            # -------------------------------------------------

            cleaned_lines = []
            previous_blank = False

            for line in lines:

                if line == "":

                    if not previous_blank:
                        cleaned_lines.append("")

                    previous_blank = True

                else:

                    cleaned_lines.append(line)
                    previous_blank = False

            text = "\n".join(cleaned_lines).strip()

            # =================================================
            # TXT 檔案路徑
            # =================================================

            txt_file = html_file.with_suffix(".txt")

            # =================================================
            # 儲存 TXT
            # =================================================

            with open(
                txt_file,
                "w",
                encoding="utf-8"
            ) as f:

                f.write(text)

            success_count += 1

            print(f"  ✓ {html_file.name} → {txt_file.name}")

        except Exception as e:

            error_count += 1

            print(f"  ✗ {html_file.name}")
            print(f"    錯誤：{e}")

# ============================================================
# 5. 最終統計
# ============================================================

print("\n")
print("=" * 70)
print("                 HTML → TXT 轉換完成")
print("=" * 70)

print(f"處理範圍：明史_卷{START_VOLUME} ～ 明史_卷{END_VOLUME}")
print(f"應處理資料夾：{END_VOLUME - START_VOLUME + 1}")
print(f"找到資料夾：{total_folders}")
print(f"找不到資料夾：{missing_folders}")
print(f"HTML 檔案總數：{total_html}")
print(f"成功轉換：{success_count}")
print(f"轉換失敗：{error_count}")

print("\nTXT 儲存位置：")
print(base_path)
```

## Chinese

```python
# ==========================================================
# Chinese Event Name Cosine Similarity
# TF-IDF Character N-gram
# 保留 event_id
# ==========================================================

import os
import json
import glob
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# ==========================================================
# 1. Settings
# ==========================================================

folder_path = "/content/"

# Character n-gram
# 中文例如：
# 始作浮梁
# 浮梁建橋
# 建造浮梁橋
NGRAM_RANGE = (1, 3)


# ==========================================================
# 2. Find JSON files
# ==========================================================

json_files = sorted(
    glob.glob(
        os.path.join(folder_path, "*.json")
    )
)

print("=" * 70)
print("JSON files")
print("=" * 70)

print(f"找到 {len(json_files)} 個 JSON 檔案")

for file_path in json_files:
    print(" -", os.path.basename(file_path))


# ==========================================================
# 3. Read JSON and extract Chinese events
#
# 中文 JSON 結構：
#
# {
#     "topic_record": {
#         "sources": [...],
#         "events": [...]
#     }
# }
# ==========================================================

events = []

for file_path in json_files:

    file_name = os.path.basename(file_path)

    with open(
        file_path,
        "r",
        encoding="utf-8"
    ) as f:

        data = json.load(f)


    # ------------------------------------------------------
    # topic_record
    # ------------------------------------------------------

    topic_record = data.get(
        "topic_record",
        {}
    )


    # ------------------------------------------------------
    # events
    # ------------------------------------------------------

    event_list = topic_record.get(
        "events",
        []
    )


    print(
        f"\n{file_name}: "
        f"{len(event_list)} events"
    )


    # ------------------------------------------------------
    # Extract event_id + event_name
    # ------------------------------------------------------

    for event in event_list:

        event_id = event.get(
            "event_id"
        )

        event_name = event.get(
            "event_name"
        )


        if event_name is None:
            continue


        event_name = str(
            event_name
        ).strip()


        if event_name == "":
            continue


        events.append({

            "event_id":
                event_id,

            "event_name":
                event_name,

            "source_file":
                file_name
        })


# ==========================================================
# 4. DataFrame
# ==========================================================

events_df = pd.DataFrame(
    events
)


print("\n" + "=" * 70)
print("Event data")
print("=" * 70)

print(
    f"Event 總數："
    f"{len(events_df):,}"
)

print(
    f"不同 event_name："
    f"{events_df['event_name'].nunique():,}"
)

print(
    f"不同 event_id："
    f"{events_df['event_id'].nunique():,}"
)


display(
    events_df.head(20)
)


# ==========================================================
# 5. Unique Events
#
# 不只用 event_name 去重
#
# 以 event_id + event_name 為唯一事件
#
# 避免：
# 不同 event_id
# 但 event_name 相同
# 被錯誤合併
# ==========================================================

unique_events_df = (
    events_df[
        [
            "event_id",
            "event_name",
            "source_file"
        ]
    ]
    .drop_duplicates(
        subset=[
            "event_id",
            "event_name"
        ]
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("Unique Events")
print("=" * 70)

print(
    f"Unique Event 數："
    f"{len(unique_events_df):,}"
)


display(
    unique_events_df
)


# ==========================================================
# 6. TF-IDF
# ==========================================================

vectorizer = TfidfVectorizer(

    # Character-level
    analyzer="char",

    # 中文 Character N-gram
    ngram_range=NGRAM_RANGE,

    # 保留設定
    lowercase=True,

    min_df=1
)


tfidf_matrix = vectorizer.fit_transform(
    unique_events_df["event_name"]
)


print("\n" + "=" * 70)
print("TF-IDF")
print("=" * 70)

print(
    "TF-IDF matrix shape:",
    tfidf_matrix.shape
)


# ==========================================================
# 7. Cosine Similarity
# ==========================================================

similarity_matrix = cosine_similarity(
    tfidf_matrix
)


# ==========================================================
# 8. Similarity Matrix
#
# 使用 event_id 作為 index / columns
# ==========================================================

similarity_df = pd.DataFrame(
    similarity_matrix,
    index=unique_events_df["event_id"],
    columns=unique_events_df["event_id"]
)


print("\n" + "=" * 70)
print("Cosine Similarity Matrix")
print("=" * 70)

display(
    similarity_df.round(4)
)


# ==========================================================
# 9. Pairwise Similarity
#
# 保留：
# event_id_1
# event_name_1
# event_id_2
# event_name_2
# cosine_similarity
# ==========================================================

pairwise_results = []


for i in range(
    len(unique_events_df)
):

    for j in range(
        i + 1,
        len(unique_events_df)
    ):

        pairwise_results.append({

            "event_id_1":
                unique_events_df.loc[
                    i,
                    "event_id"
                ],

            "event_name_1":
                unique_events_df.loc[
                    i,
                    "event_name"
                ],

            "event_id_2":
                unique_events_df.loc[
                    j,
                    "event_id"
                ],

            "event_name_2":
                unique_events_df.loc[
                    j,
                    "event_name"
                ],

            "cosine_similarity":
                similarity_matrix[i, j]
        })


pairwise_df = pd.DataFrame(
    pairwise_results
)


# ==========================================================
# 10. Sort Similarity
# ==========================================================

pairwise_df = (
    pairwise_df
    .sort_values(
        "cosine_similarity",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("Pairwise Cosine Similarity")
print("=" * 70)

display(
    pairwise_df.round(4)
)


# ==========================================================
# 11. Save original Event data
# ==========================================================

events_path = (
    "/content/"
    "chinese_event_events.csv"
)


events_df.to_csv(
    events_path,
    index=False,
    encoding="utf-8-sig"
)


# ==========================================================
# 12. Save Event mapping
# ==========================================================

mapping_path = (
    "/content/"
    "chinese_event_mapping.csv"
)


unique_events_df.to_csv(
    mapping_path,
    index=False,
    encoding="utf-8-sig"
)


# ==========================================================
# 13. Save Similarity Matrix
# ==========================================================

matrix_path = (
    "/content/"
    "chinese_event_cosine_similarity_matrix.csv"
)


similarity_df.to_csv(
    matrix_path,
    encoding="utf-8-sig"
)


# ==========================================================
# 14. Save Pairwise Similarity
# ==========================================================

pairwise_path = (
    "/content/"
    "chinese_event_cosine_similarity_pairwise.csv"
)


pairwise_df.to_csv(
    pairwise_path,
    index=False,
    encoding="utf-8-sig"
)


# ==========================================================
# 15. 完成
# ==========================================================

print("\n" + "=" * 70)
print("完成")
print("=" * 70)

print(
    "\nEvent data：",
    events_path
)

print(
    "Event mapping：",
    mapping_path
)

print(
    "Similarity matrix：",
    matrix_path
)

print(
    "Pairwise results：",
    pairwise_path
)
```

## English

```python
# ============================================================
# English Event Name Cosine Similarity
# 保留 event_id
# ============================================================

import os
import json
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# ============================================================
# 1. 指定英文 JSON 檔案
# ============================================================

folder_path = "/content/"

json_files = [
    "annotations_enc_rotary_en_1319843498.json",
    "annotations_enc_rotary_en_1319854677.json",
    "po_en_1.json",
    "po_en_1669.json"
]


# ============================================================
# 2. 讀取 JSON
# ============================================================

events = []

for filename in json_files:

    file_path = os.path.join(folder_path, filename)

    if not os.path.exists(file_path):
        print(f"⚠ 找不到檔案：{filename}")
        continue

    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    # --------------------------------------------------------
    # JSON 最外層可能是：
    #
    # "ENC/rotary/eng"
    #
    # 或其他 document key
    # --------------------------------------------------------

    for document_key, document_data in data.items():

        if not isinstance(document_data, dict):
            continue

        event_list = document_data.get("events", [])

        for event in event_list:

            event_id = event.get("event_id")
            event_name = event.get("event_name")

            if event_name is None:
                continue

            event_name = str(event_name).strip()

            if event_name == "":
                continue

            events.append({
                "event_id": event_id,
                "event_name": event_name,
                "source_file": filename,
                "document_key": document_key
            })


# ============================================================
# 3. 建立原始 Event DataFrame
# ============================================================

events_df = pd.DataFrame(events)

print("=" * 70)
print("資料讀取結果")
print("=" * 70)

print(f"JSON 檔案數：{len(json_files)}")
print(f"Event 總數：{len(events_df)}")

if len(events_df) > 0:

    print(
        f"不同 event_name 數："
        f"{events_df['event_name'].nunique()}"
    )

    print("\n各檔案 Event 數量：")
    print(
        events_df["source_file"]
        .value_counts()
    )


# ============================================================
# 4. 保留 event_id 的 Unique Event
#
# 注意：
# 同一 event_name 如果對應不同 event_id，
# 這裡仍然會保留不同 event。
#
# 只有 event_id + event_name 完全相同時才去除重複。
# ============================================================

unique_events_df = (
    events_df[
        [
            "event_id",
            "event_name",
            "source_file",
            "document_key"
        ]
    ]
    .drop_duplicates(
        subset=["event_id", "event_name"]
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("Unique Events")
print("=" * 70)

print(
    f"Unique Event 數："
    f"{len(unique_events_df)}"
)

display(unique_events_df)


# ============================================================
# 5. TF-IDF
# ============================================================

vectorizer = TfidfVectorizer(
    analyzer="word",
    ngram_range=(1, 2),
    lowercase=True,
    stop_words="english",
    min_df=1
)

tfidf_matrix = vectorizer.fit_transform(
    unique_events_df["event_name"]
)


print("\n" + "=" * 70)
print("TF-IDF")
print("=" * 70)

print(
    f"TF-IDF Matrix Shape："
    f"{tfidf_matrix.shape}"
)


# ============================================================
# 6. Cosine Similarity
# ============================================================

similarity_matrix = cosine_similarity(
    tfidf_matrix
)


# ============================================================
# 7. 建立 Pairwise Similarity
#
# CSV 會保留：
#
# event_id_1
# event_name_1
# event_id_2
# event_name_2
# cosine_similarity
# ============================================================

pairwise_results = []

n_events = len(unique_events_df)

for i in range(n_events):

    for j in range(i + 1, n_events):

        pairwise_results.append({

            "event_id_1":
                unique_events_df.loc[i, "event_id"],

            "event_name_1":
                unique_events_df.loc[i, "event_name"],

            "event_id_2":
                unique_events_df.loc[j, "event_id"],

            "event_name_2":
                unique_events_df.loc[j, "event_name"],

            "cosine_similarity":
                similarity_matrix[i, j]
        })


pairwise_df = pd.DataFrame(
    pairwise_results
)


# ============================================================
# 8. 由高到低排序
# ============================================================

pairwise_df = (
    pairwise_df
    .sort_values(
        "cosine_similarity",
        ascending=False
    )
    .reset_index(drop=True)
)


# ============================================================
# 9. 顯示最高相似度
# ============================================================

print("\n" + "=" * 70)
print("Top 30 Most Similar Event Pairs")
print("=" * 70)

display(
    pairwise_df
    .head(30)
    .round(4)
)


# ============================================================
# 10. 建立完整 Similarity Matrix
#
# index / columns 使用 event_id
# ============================================================

similarity_df = pd.DataFrame(
    similarity_matrix,
    index=unique_events_df["event_id"],
    columns=unique_events_df["event_id"]
)


print("\n" + "=" * 70)
print("Cosine Similarity Matrix")
print("=" * 70)

display(
    similarity_df.round(4)
)


# ============================================================
# 11. 建立 Event 對照表
#
# event_id → event_name
# ============================================================

event_mapping_df = unique_events_df[
    [
        "event_id",
        "event_name",
        "source_file",
        "document_key"
    ]
].copy()


# ============================================================
# 12. 儲存 CSV
# ============================================================

events_output = (
    "/content/english_events.csv"
)

mapping_output = (
    "/content/english_event_mapping.csv"
)

matrix_output = (
    "/content/english_event_cosine_similarity_matrix.csv"
)

pairwise_output = (
    "/content/english_event_cosine_similarity_pairwise.csv"
)


# 原始 Event
events_df.to_csv(
    events_output,
    index=False,
    encoding="utf-8-sig"
)


# Event ID 對照表
event_mapping_df.to_csv(
    mapping_output,
    index=False,
    encoding="utf-8-sig"
)


# Similarity Matrix
similarity_df.to_csv(
    matrix_output,
    encoding="utf-8-sig"
)


# Pairwise Similarity
pairwise_df.to_csv(
    pairwise_output,
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 13. 完成
# ============================================================

print("\n" + "=" * 70)
print("完成")
print("=" * 70)

print(f"✓ 原始 Event CSV：")
print(f"  {events_output}")

print(f"\n✓ Event ID 對照表：")
print(f"  {mapping_output}")

print(f"\n✓ Cosine Similarity Matrix：")
print(f"  {matrix_output}")

print(f"\n✓ Pairwise Cosine Similarity：")
print(f"  {pairwise_output}")
```

## Annotations v.s. LLM

```python
# ==========================================================
# Event Name Cosine Similarity
#
# 比較：
# 1. 中、英文 JSON 檔案中的 event_name
# 2. extraction_events_final_schema.json 中的 event_name
#
# 方法：
# TF-IDF Character N-gram
# Cosine Similarity
#
# 保留：
# source_event_id
# source_event_name
# schema_event_id
# schema_event_name
# ==========================================================

import os
import json
import glob
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# ==========================================================
# 1. Settings
# ==========================================================

folder_path = "/content/"

# Schema file
schema_path = (
    "/content/extraction_events_final_schema.json"
)

# Character N-gram
#
# 可以同時處理：
# 中文：
#   橋燬
#   橋落成
#
# 英文：
#   Bridge was destroyed
#   Construction of the bridge
#
# 中英混合：
#   Bridge construction 造橋
#
NGRAM_RANGE = (1, 3)


# ==========================================================
# 2. Read source JSON files
#
# 排除 extraction_events_final_schema.json
# ==========================================================

json_files = sorted(
    [
        f
        for f in glob.glob(
            os.path.join(
                folder_path,
                "*.json"
            )
        )
        if os.path.abspath(f)
        != os.path.abspath(schema_path)
    ]
)


print("=" * 70)
print("Source JSON files")
print("=" * 70)

print(
    f"找到 {len(json_files)} 個來源 JSON 檔案"
)

for file_path in json_files:
    print(
        " -",
        os.path.basename(file_path)
    )


# ==========================================================
# 3. Extract events from source JSON files
#
# 支援：
#
# 中文 JSON：
# {
#     "topic_record": {
#         "events": [...]
#     }
# }
#
# 英文 JSON：
# {
#     "ENC/rotary/eng": {
#         "events": [...]
#     }
# }
# ==========================================================

source_events = []


for file_path in json_files:

    file_name = os.path.basename(
        file_path
    )

    with open(
        file_path,
        "r",
        encoding="utf-8"
    ) as f:

        data = json.load(f)


    file_event_count = 0


    # ------------------------------------------------------
    # Case 1
    # topic_record -> events
    # ------------------------------------------------------

    if isinstance(data, dict):

        topic_record = data.get(
            "topic_record"
        )

        if isinstance(
            topic_record,
            dict
        ):

            event_list = (
                topic_record.get(
                    "events",
                    []
                )
            )

            for event in event_list:

                if not isinstance(
                    event,
                    dict
                ):
                    continue

                event_id = event.get(
                    "event_id"
                )

                event_name = event.get(
                    "event_name"
                )

                if event_name is None:
                    continue

                event_name = str(
                    event_name
                ).strip()

                if event_name == "":
                    continue

                source_events.append({

                    "source_event_id":
                        event_id,

                    "source_event_name":
                        event_name,

                    "source_file":
                        file_name

                })

                file_event_count += 1


    # ------------------------------------------------------
    # Case 2
    # 其他最外層 key -> events
    #
    # 例如：
    #
    # "ENC/rotary/eng": {
    #     "events": [...]
    # }
    # ------------------------------------------------------

    if isinstance(data, dict):

        for document_key, document_data in data.items():

            # topic_record 已經處理
            if document_key == "topic_record":
                continue

            if not isinstance(
                document_data,
                dict
            ):
                continue

            event_list = (
                document_data.get(
                    "events"
                )
            )

            if not isinstance(
                event_list,
                list
            ):
                continue


            for event in event_list:

                if not isinstance(
                    event,
                    dict
                ):
                    continue

                event_id = event.get(
                    "event_id"
                )

                event_name = event.get(
                    "event_name"
                )

                if event_name is None:
                    continue

                event_name = str(
                    event_name
                ).strip()

                if event_name == "":
                    continue

                source_events.append({

                    "source_event_id":
                        event_id,

                    "source_event_name":
                        event_name,

                    "source_file":
                        file_name

                })

                file_event_count += 1


    print(
        f"{file_name}: "
        f"{file_event_count} events"
    )


# ==========================================================
# 4. Source Event DataFrame
# ==========================================================

source_df = pd.DataFrame(
    source_events
)


print("\n" + "=" * 70)
print("Source Event Data")
print("=" * 70)

print(
    f"來源 Event 總數："
    f"{len(source_df):,}"
)

print(
    f"不同 source event_id："
    f"{source_df['source_event_id'].nunique():,}"
)

print(
    f"不同 source event_name："
    f"{source_df['source_event_name'].nunique():,}"
)

display(
    source_df.head(20)
)


# ==========================================================
# 5. Read extraction_events_final_schema.json
# ==========================================================

print("\n" + "=" * 70)
print("Reading extraction_events_final_schema.json")
print("=" * 70)


with open(
    schema_path,
    "r",
    encoding="utf-8"
) as f:

    schema_data = json.load(f)


schema_events = schema_data.get(
    "events",
    []
)


print(
    f"Schema Event 總數："
    f"{len(schema_events):,}"
)


# ==========================================================
# 6. Extract Schema Events
# ==========================================================

schema_event_records = []


for event in schema_events:

    if not isinstance(
        event,
        dict
    ):
        continue

    event_id = event.get(
        "event_id"
    )

    event_name = event.get(
        "event_name"
    )

    if event_name is None:
        continue

    event_name = str(
        event_name
    ).strip()

    if event_name == "":
        continue

    schema_event_records.append({

        "schema_event_id":
            event_id,

        "schema_event_name":
            event_name

    })


schema_df = pd.DataFrame(
    schema_event_records
)


print(
    f"有效 Schema Event："
    f"{len(schema_df):,}"
)

print(
    f"不同 schema event_id："
    f"{schema_df['schema_event_id'].nunique():,}"
)

print(
    f"不同 schema event_name："
    f"{schema_df['schema_event_name'].nunique():,}"
)


display(
    schema_df.head(20)
)


# ==========================================================
# 7. TF-IDF
#
# 重要：
#
# source event + schema event
# 一起建立 TF-IDF vocabulary
#
# 這樣兩邊會使用同一個向量空間。
# ==========================================================

all_event_names = pd.concat(
    [
        source_df[
            "source_event_name"
        ],
        schema_df[
            "schema_event_name"
        ]
    ],
    ignore_index=True
)


vectorizer = TfidfVectorizer(

    # Character-level
    analyzer="char",

    # Chinese / English 都可以
    ngram_range=NGRAM_RANGE,

    # 英文大小寫不區分
    lowercase=True,

    min_df=1
)


tfidf_all = vectorizer.fit_transform(
    all_event_names
)


# ==========================================================
# 8. Separate source / schema vectors
# ==========================================================

source_count = len(
    source_df
)

schema_count = len(
    schema_df
)


source_tfidf = (
    tfidf_all[
        :source_count
    ]
)


schema_tfidf = (
    tfidf_all[
        source_count:
    ]
)


print("\n" + "=" * 70)
print("TF-IDF")
print("=" * 70)

print(
    "Source TF-IDF shape:",
    source_tfidf.shape
)

print(
    "Schema TF-IDF shape:",
    schema_tfidf.shape
)


# ==========================================================
# 9. Cosine Similarity
#
# 每一個 source event
# 對每一個 schema event
#
# 例如：
#
# source_event_1
#       ↓
# schema_event_1
# schema_event_2
# schema_event_3
# ...
#
# source_event_2
#       ↓
# schema_event_1
# schema_event_2
# ...
# ==========================================================

similarity_matrix = cosine_similarity(
    source_tfidf,
    schema_tfidf
)


print("\n" + "=" * 70)
print("Cosine Similarity")
print("=" * 70)

print(
    "Similarity matrix shape:",
    similarity_matrix.shape
)


# ==========================================================
# 10. 建立 Pairwise Results
#
# 保留雙方 event_id
# ==========================================================

pairwise_results = []


for i in range(
    source_count
):

    for j in range(
        schema_count
    ):

        pairwise_results.append({

            "source_event_id":
                source_df.iloc[
                    i
                ]["source_event_id"],

            "source_event_name":
                source_df.iloc[
                    i
                ]["source_event_name"],

            "source_file":
                source_df.iloc[
                    i
                ]["source_file"],

            "schema_event_id":
                schema_df.iloc[
                    j
                ]["schema_event_id"],

            "schema_event_name":
                schema_df.iloc[
                    j
                ]["schema_event_name"],

            "cosine_similarity":
                similarity_matrix[
                    i,
                    j
                ]

        })


pairwise_df = pd.DataFrame(
    pairwise_results
)


# ==========================================================
# 11. Sort by similarity
# ==========================================================

pairwise_df = (
    pairwise_df
    .sort_values(
        "cosine_similarity",
        ascending=False
    )
    .reset_index(drop=True)
)


# ==========================================================
# 12. Display Top 50
# ==========================================================

print("\n" + "=" * 70)
print("Top 50 Similar Event Pairs")
print("=" * 70)

display(
    pairwise_df
    .head(50)
    .round(4)
)


# ==========================================================
# 13. 每個 Source Event 找最高相似的 Schema Event
#
# 這個結果通常比完整 pairwise 更適合後續人工檢查。
# ==========================================================

best_match_df = (
    pairwise_df
    .sort_values(
        "cosine_similarity",
        ascending=False
    )
    .drop_duplicates(
        subset=[
            "source_event_id"
        ]
    )
    .sort_values(
        "source_event_id"
    )
    .reset_index(drop=True)
)


print("\n" + "=" * 70)
print("Best Schema Match for Each Source Event")
print("=" * 70)

display(
    best_match_df
)


# ==========================================================
# 14. Save source events
# ==========================================================

source_output = (
    "/content/"
    "source_events_with_event_id.csv"
)


source_df.to_csv(
    source_output,
    index=False,
    encoding="utf-8-sig"
)


# ==========================================================
# 15. Save schema events
# ==========================================================

schema_output = (
    "/content/"
    "schema_events_with_event_id.csv"
)


schema_df.to_csv(
    schema_output,
    index=False,
    encoding="utf-8-sig"
)


# ==========================================================
# 16. Save complete pairwise similarity
# ==========================================================

pairwise_output = (
    "/content/"
    "source_vs_schema_event_cosine_similarity.csv"
)


pairwise_df.to_csv(
    pairwise_output,
    index=False,
    encoding="utf-8-sig"
)


# ==========================================================
# 17. Save best match
# ==========================================================

best_match_output = (
    "/content/"
    "source_vs_schema_best_event_match.csv"
)


best_match_df.to_csv(
    best_match_output,
    index=False,
    encoding="utf-8-sig"
)


# ==========================================================
# 18. 完成
# ==========================================================

print("\n" + "=" * 70)
print("完成")
print("=" * 70)

print(
    "\nSource Events："
)

print(
    source_output
)

print(
    "\nSchema Events："
)

print(
    schema_output
)

print(
    "\nComplete Pairwise Similarity："
)

print(
    pairwise_output
)

print(
    "\nBest Match："
)

print(
    best_match_output
)
```

```python
# ============================================================
# Event Name Semantic Similarity
# Chinese: BAAI/bge-large-zh-v1.5
# English: BAAI/bge-large-en-v1.5
#
# Input:
#   /content/*.json
#   /content/extraction_events_final_schema.json
#
# Output:
#   /content/chinese_event_similarity_pairwise.csv
#   /content/chinese_event_similarity_top5.csv
#   /content/chinese_event_similarity_top1.csv
#   /content/english_event_similarity_pairwise.csv
#   /content/english_event_similarity_top5.csv
#   /content/english_event_similarity_top1.csv
# ============================================================

# ------------------------------------------------------------
# 1. Install packages
# ------------------------------------------------------------
!pip -q install -U sentence-transformers transformers accelerate

# ------------------------------------------------------------
# 2. Import
# ------------------------------------------------------------
import os
import glob
import json
import re
import numpy as np
import pandas as pd

from tqdm.auto import tqdm
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity


# ============================================================
# 3. Settings
# ============================================================

SOURCE_FOLDER = "/content/"
SCHEMA_PATH = "/content/extraction_events_final_schema.json"

CHINESE_MODEL_NAME = "BAAI/bge-large-zh-v1.5"
ENGLISH_MODEL_NAME = "BAAI/bge-large-en-v1.5"

TOP_K = 5

# 每次 encode 的 batch size
BATCH_SIZE = 32

# 是否使用 GPU
DEVICE = "cuda" if __import__("torch").cuda.is_available() else "cpu"

print("=" * 70)
print("Settings")
print("=" * 70)
print("Chinese model :", CHINESE_MODEL_NAME)
print("English model :", ENGLISH_MODEL_NAME)
print("Device        :", DEVICE)
print("Top-K         :", TOP_K)
print("=" * 70)


# ============================================================
# 4. 判斷文字語言
# ============================================================

def detect_language(text):
    """
    簡單判斷 event_name 是中文或英文。

    規則：
    - 中文字元比例較高 → zh
    - 否則 → en
    """

    if not isinstance(text, str):
        return "unknown"

    text = text.strip()

    if not text:
        return "unknown"

    chinese_chars = re.findall(r'[\u4e00-\u9fff]', text)
    english_chars = re.findall(r'[A-Za-z]', text)

    zh_count = len(chinese_chars)
    en_count = len(english_chars)

    if zh_count > en_count:
        return "zh"

    elif en_count > zh_count:
        return "en"

    # 如果兩者一樣，依是否出現中文判斷
    elif zh_count > 0:
        return "zh"

    else:
        return "en"


# ============================================================
# 5. 讀取 source JSON
# ============================================================

def extract_events_from_json(data, filename):
    """
    支援目前的兩種 JSON 結構：

    中文：
    {
        "topic_record": {
            "events": [...]
        }
    }

    英文：
    {
        "ENC/rotary/eng": {
            "events": [...]
        }
    }
    """

    records = []

    # --------------------------------------------------------
    # Case 1: Chinese
    # topic_record -> events
    # --------------------------------------------------------
    if isinstance(data, dict):

        topic_record = data.get("topic_record")

        if isinstance(topic_record, dict):

            events = topic_record.get("events", [])

            if isinstance(events, list):

                for event in events:

                    if not isinstance(event, dict):
                        continue

                    event_id = event.get("event_id")
                    event_name = event.get("event_name")

                    if event_id is None or event_name is None:
                        continue

                    records.append({
                        "source_event_id": str(event_id),
                        "source_event_name": str(event_name),
                        "source_file": filename,
                        "language": detect_language(str(event_name))
                    })

                return records

    # --------------------------------------------------------
    # Case 2: English / other structure
    #
    # {
    #     "ENC/rotary/eng": {
    #         "events": [...]
    #     }
    # }
    # --------------------------------------------------------
    if isinstance(data, dict):

        for key, value in data.items():

            if not isinstance(value, dict):
                continue

            events = value.get("events")

            if not isinstance(events, list):
                continue

            for event in events:

                if not isinstance(event, dict):
                    continue

                event_id = event.get("event_id")
                event_name = event.get("event_name")

                if event_id is None or event_name is None:
                    continue

                records.append({
                    "source_event_id": str(event_id),
                    "source_event_name": str(event_name),
                    "source_file": filename,
                    "language": detect_language(str(event_name))
                })

    return records


# ============================================================
# 6. 找到 source JSON files
# ============================================================

json_files = sorted(glob.glob(os.path.join(SOURCE_FOLDER, "*.json")))

# 排除 schema
json_files = [
    f for f in json_files
    if os.path.abspath(f) != os.path.abspath(SCHEMA_PATH)
]

print("\nSource JSON files:")
for f in json_files:
    print(" -", os.path.basename(f))

print("\nSchema:")
print(" -", SCHEMA_PATH)


# ============================================================
# 7. 讀取 source events
# ============================================================

source_records = []

for filepath in json_files:

    filename = os.path.basename(filepath)

    try:

        with open(filepath, "r", encoding="utf-8") as f:
            data = json.load(f)

        records = extract_events_from_json(
            data,
            filename
        )

        source_records.extend(records)

        print(
            f"{filename:<55} "
            f"{len(records):>8,} events"
        )

    except Exception as e:

        print(
            f"[ERROR] {filename}: {e}"
        )


source_df = pd.DataFrame(source_records)


# ============================================================
# 8. 去除完全重複的 source event
#
# 注意：
# 不可以只用 event_name 去重。
#
# 因為：
# 同一 event_name 可能有不同 event_id。
# ============================================================

if len(source_df) > 0:

    source_df = (
        source_df
        .drop_duplicates(
            subset=[
                "source_event_id",
                "source_event_name",
                "source_file"
            ]
        )
        .reset_index(drop=True)
    )


# ============================================================
# 9. 讀取 schema
# ============================================================

print("\n" + "=" * 70)
print("Loading schema")
print("=" * 70)

with open(SCHEMA_PATH, "r", encoding="utf-8") as f:
    schema_data = json.load(f)


schema_events = schema_data.get("events", [])

schema_records = []

for event in schema_events:

    if not isinstance(event, dict):
        continue

    event_id = event.get("event_id")
    event_name = event.get("event_name")

    if event_id is None or event_name is None:
        continue

    schema_records.append({
        "schema_event_id": str(event_id),
        "schema_event_name": str(event_name),
        "language": detect_language(str(event_name))
    })


schema_df = pd.DataFrame(schema_records)


# ------------------------------------------------------------
# Schema 去除完全重複
# ------------------------------------------------------------

if len(schema_df) > 0:

    schema_df = (
        schema_df
        .drop_duplicates(
            subset=[
                "schema_event_id",
                "schema_event_name"
            ]
        )
        .reset_index(drop=True)
    )


# ============================================================
# 10. 顯示資料統計
# ============================================================

print("\n" + "=" * 70)
print("Event Statistics")
print("=" * 70)

print("\nSOURCE EVENTS")
print(source_df["language"].value_counts(dropna=False))

print("\nSCHEMA EVENTS")
print(schema_df["language"].value_counts(dropna=False))

print("\nSource total :", len(source_df))
print("Schema total :", len(schema_df))


# ============================================================
# 11. 分開中文 / 英文
# ============================================================

source_zh = source_df[
    source_df["language"] == "zh"
].copy()

source_en = source_df[
    source_df["language"] == "en"
].copy()


schema_zh = schema_df[
    schema_df["language"] == "zh"
].copy()

schema_en = schema_df[
    schema_df["language"] == "en"
].copy()


print("\n" + "=" * 70)
print("Language-specific data")
print("=" * 70)

print(f"Chinese source : {len(source_zh):,}")
print(f"Chinese schema : {len(schema_zh):,}")

print(f"English source : {len(source_en):,}")
print(f"English schema : {len(schema_en):,}")


# ============================================================
# 12. 載入中文模型
# ============================================================

print("\n" + "=" * 70)
print("Loading Chinese model")
print("=" * 70)

zh_model = SentenceTransformer(
    CHINESE_MODEL_NAME,
    device=DEVICE
)


# ============================================================
# 13. 載入英文模型
# ============================================================

print("\n" + "=" * 70)
print("Loading English model")
print("=" * 70)

en_model = SentenceTransformer(
    ENGLISH_MODEL_NAME,
    device=DEVICE
)


# ============================================================
# 14. Embedding function
# ============================================================

def encode_texts(model, texts, batch_size=32):

    if len(texts) == 0:
        return np.empty((0, 0))

    embeddings = model.encode(
        texts,
        batch_size=batch_size,
        show_progress_bar=True,
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    return embeddings


# ============================================================
# 15. 中文 embedding
# ============================================================

print("\n" + "=" * 70)
print("Encoding Chinese events")
print("=" * 70)

if len(source_zh) > 0 and len(schema_zh) > 0:

    zh_source_embeddings = encode_texts(
        zh_model,
        source_zh["source_event_name"].tolist(),
        BATCH_SIZE
    )

    zh_schema_embeddings = encode_texts(
        zh_model,
        schema_zh["schema_event_name"].tolist(),
        BATCH_SIZE
    )

else:

    zh_source_embeddings = np.empty((0, 0))
    zh_schema_embeddings = np.empty((0, 0))


# ============================================================
# 16. 英文 embedding
# ============================================================

print("\n" + "=" * 70)
print("Encoding English events")
print("=" * 70)

if len(source_en) > 0 and len(schema_en) > 0:

    en_source_embeddings = encode_texts(
        en_model,
        source_en["source_event_name"].tolist(),
        BATCH_SIZE
    )

    en_schema_embeddings = encode_texts(
        en_model,
        schema_en["schema_event_name"].tolist(),
        BATCH_SIZE
    )

else:

    en_source_embeddings = np.empty((0, 0))
    en_schema_embeddings = np.empty((0, 0))


# ============================================================
# 17. 中文 Cosine Similarity
# ============================================================

print("\n" + "=" * 70)
print("Calculating Chinese cosine similarity")
print("=" * 70)

if len(source_zh) > 0 and len(schema_zh) > 0:

    zh_similarity_matrix = cosine_similarity(
        zh_source_embeddings,
        zh_schema_embeddings
    )

else:

    zh_similarity_matrix = np.empty(
        (len(source_zh), len(schema_zh))
    )


# ============================================================
# 18. 英文 Cosine Similarity
# ============================================================

print("\n" + "=" * 70)
print("Calculating English cosine similarity")
print("=" * 70)

if len(source_en) > 0 and len(schema_en) > 0:

    en_similarity_matrix = cosine_similarity(
        en_source_embeddings,
        en_schema_embeddings
    )

else:

    en_similarity_matrix = np.empty(
        (len(source_en), len(schema_en))
    )


# ============================================================
# 19. 將 similarity matrix 轉成 pairwise table
# ============================================================

def similarity_to_pairwise(
    source_df,
    schema_df,
    similarity_matrix
):

    rows = []

    for i in range(len(source_df)):

        source_row = source_df.iloc[i]

        for j in range(len(schema_df)):

            schema_row = schema_df.iloc[j]

            rows.append({
                "source_event_id":
                    source_row["source_event_id"],

                "source_event_name":
                    source_row["source_event_name"],

                "source_file":
                    source_row["source_file"],

                "schema_event_id":
                    schema_row["schema_event_id"],

                "schema_event_name":
                    schema_row["schema_event_name"],

                "cosine_similarity":
                    float(similarity_matrix[i, j])
            })

    return pd.DataFrame(rows)


# ============================================================
# 20. 中文 pairwise
# ============================================================

print("\nCreating Chinese pairwise results...")

zh_pairwise_df = similarity_to_pairwise(
    source_zh,
    schema_zh,
    zh_similarity_matrix
)

if len(zh_pairwise_df) > 0:

    zh_pairwise_df = (
        zh_pairwise_df
        .sort_values(
            "cosine_similarity",
            ascending=False
        )
        .reset_index(drop=True)
    )


# ============================================================
# 21. 英文 pairwise
# ============================================================

print("Creating English pairwise results...")

en_pairwise_df = similarity_to_pairwise(
    source_en,
    schema_en,
    en_similarity_matrix
)

if len(en_pairwise_df) > 0:

    en_pairwise_df = (
        en_pairwise_df
        .sort_values(
            "cosine_similarity",
            ascending=False
        )
        .reset_index(drop=True)
    )


# ============================================================
# 22. 每個 source event 的 Top-K
# ============================================================

def get_top_k_matches(
    pairwise_df,
    k=5
):

    if len(pairwise_df) == 0:
        return pairwise_df.copy()

    return (
        pairwise_df
        .sort_values(
            [
                "source_event_id",
                "cosine_similarity"
            ],
            ascending=[
                True,
                False
            ]
        )
        .groupby(
            "source_event_id",
            group_keys=False
        )
        .head(k)
        .reset_index(drop=True)
    )


zh_top5_df = get_top_k_matches(
    zh_pairwise_df,
    TOP_K
)

en_top5_df = get_top_k_matches(
    en_pairwise_df,
    TOP_K
)


# ============================================================
# 23. 每個 source event 的 Top-1
# ============================================================

def get_top1_matches(pairwise_df):

    if len(pairwise_df) == 0:
        return pairwise_df.copy()

    return (
        pairwise_df
        .sort_values(
            "cosine_similarity",
            ascending=False
        )
        .drop_duplicates(
            subset=["source_event_id"]
        )
        .sort_values(
            "source_event_id"
        )
        .reset_index(drop=True)
    )


zh_top1_df = get_top1_matches(
    zh_pairwise_df
)

en_top1_df = get_top1_matches(
    en_pairwise_df
)


# ============================================================
# 24. 儲存 CSV
# ============================================================

OUTPUT_FILES = {
    "zh_pairwise":
        "/content/chinese_event_similarity_pairwise.csv",

    "zh_top5":
        "/content/chinese_event_similarity_top5.csv",

    "zh_top1":
        "/content/chinese_event_similarity_top1.csv",

    "en_pairwise":
        "/content/english_event_similarity_pairwise.csv",

    "en_top5":
        "/content/english_event_similarity_top5.csv",

    "en_top1":
        "/content/english_event_similarity_top1.csv",
}


zh_pairwise_df.to_csv(
    OUTPUT_FILES["zh_pairwise"],
    index=False,
    encoding="utf-8-sig"
)

zh_top5_df.to_csv(
    OUTPUT_FILES["zh_top5"],
    index=False,
    encoding="utf-8-sig"
)

zh_top1_df.to_csv(
    OUTPUT_FILES["zh_top1"],
    index=False,
    encoding="utf-8-sig"
)

en_pairwise_df.to_csv(
    OUTPUT_FILES["en_pairwise"],
    index=False,
    encoding="utf-8-sig"
)

en_top5_df.to_csv(
    OUTPUT_FILES["en_top5"],
    index=False,
    encoding="utf-8-sig"
)

en_top1_df.to_csv(
    OUTPUT_FILES["en_top1"],
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 25. 顯示結果
# ============================================================

print("\n" + "=" * 70)
print("Completed!")
print("=" * 70)

print("\nOutput files:")

for name, path in OUTPUT_FILES.items():

    print(
        f"{name:<15} "
        f"{path}"
    )


# ============================================================
# 26. 顯示 Top-1 結果
# ============================================================

print("\n" + "=" * 70)
print("Chinese Top-1 Matches")
print("=" * 70)

if len(zh_top1_df) > 0:

    display(
        zh_top1_df[
            [
                "source_event_id",
                "source_event_name",
                "schema_event_id",
                "schema_event_name",
                "cosine_similarity"
            ]
        ].head(20)
    )

else:

    print("No Chinese matching results.")


print("\n" + "=" * 70)
print("English Top-1 Matches")
print("=" * 70)

if len(en_top1_df) > 0:

    display(
        en_top1_df[
            [
                "source_event_id",
                "source_event_name",
                "schema_event_id",
                "schema_event_name",
                "cosine_similarity"
            ]
        ].head(20)
    )

else:

    print("No English matching results.")


# ============================================================
# 27. 顯示高相似度配對
# ============================================================

print("\n" + "=" * 70)
print("Chinese Top-5 Examples")
print("=" * 70)

if len(zh_top5_df) > 0:

    display(
        zh_top5_df[
            [
                "source_event_id",
                "source_event_name",
                "schema_event_id",
                "schema_event_name",
                "cosine_similarity"
            ]
        ].head(25)
    )


print("\n" + "=" * 70)
print("English Top-5 Examples")
print("=" * 70)

if len(en_top5_df) > 0:

    display(
        en_top5_df[
            [
                "source_event_id",
                "source_event_name",
                "schema_event_id",
                "schema_event_name",
                "cosine_similarity"
            ]
        ].head(25)
    )
```
